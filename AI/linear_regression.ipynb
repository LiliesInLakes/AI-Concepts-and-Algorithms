```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)
x= np.random.rand(50, 1)*100
x_test= np.random.rand(20, 1)*100
y= 3.5*x+ np.random.rand(50, 1)*40
y_test= 3.5*x_test+ np.random.rand(20, 1)*20
# print(x)
# print(y)
```

```python
model= LinearRegression()
model.fit(x, y)
```

```python
y_pred= model.predict(x)
print(y_pred)
```

```python
plt.scatter(x, y, color= 'red', label= 'data_points')
plt.scatter(x, y_pred, color='green')
plt.plot(x, y_pred)
```

```python
slope= model.coef_[0][0]
intercept= model.intercept_[0]
print(slope, intercept)
```
