```python
import numpy as np
import math
```

```python
docs= ["The cow jumped over the moon"
,"O'Leary's cow kicked the lamp"
,"The kicked lamp started a fire"
,"The cow on fire"]
print(docs[1])
```

```python
vocb = {}

for d in docs:
    lowered= d.lower()
    tok = lowered.split()
    for key in tok:
        ret = vocb.get(key, 0)
        if ret == 0:
            vocb.update({key : 1})
        else:
            vocb[key]+=1
print(vocb) 
```

```python
mtrix= []
for word in vocb:
    row = [0, 0, 0, 0]
    i =-1
    for d in docs:
        i+=1
        lowered = d.lower()
        tok = lowered.split()
        for key in tok:
            if key == word:
                row[i] +=1
    mtrix.append(row)
print(mtrix)
```

```python
#columns 
for i in range(0, 4):
    column = []
    for j in range(0, len(mtrix)):
        column.append(mtrix[j][i])
    for k in range(i+1, 4):
        column2 = []
        for j2 in range(0, len(mtrix)):
            column2.append(mtrix[j2][k])
        sum_sq = sum((a - b) ** 2 for a, b in zip(column, column2))
        distance = math.sqrt(sum_sq)
        print(f"d{i} and d{k}: {distance}")
```

```python
#cosine similarity
for i in range(0, 4):
    column = []
    for j in range(0, len(mtrix)):
        column.append(mtrix[j][i])
    for k in range(i+1, 4):
        column2 = []
        for j2 in range(0, len(mtrix)):
            column2.append(mtrix[j2][k])
        dotp = sum((a *b) for a, b in zip(column, column2))
        c1= sum(a**2 for a in column)
        c2= sum(b**2 for b in column2)
        distance = dotp/(math.sqrt(c1)*math.sqrt(c2))
        print(f"d{i} and d{k}: {distance}")
```

```python
#tf idf matrix
matrix = []
for i in range(0, len(mtrix)):
    row = []
    df = 0
    for k in range(0, len(mtrix[0])):
        if mtrix[i][k] != 0:
            df+=1
    for j in range(0, len(mtrix[0])):
        tf = mtrix[i][j]
        idf= math.log(len(mtrix)/df) if df > 0 else 0.0
        row.append(tf*idf)
    matrix.append(row)
print(matrix)
```

```python
#cosine similarity using tfidf vectors
for i in range(0, 4):
    column = []
    for j in range(0, len(matrix)):
        column.append(matrix[j][i])
    for k in range(i+1, 4):
        column2 = []
        for j2 in range(0, len(mtrix)):
            column2.append(matrix[j2][k])
        dotp = sum((a *b) for a, b in zip(column, column2))
        c1= sum(a**2 for a in column)
        c2= sum(b**2 for b in column2)
        distance = dotp/(math.sqrt(c1)*math.sqrt(c2))
        print(f"d{i} and d{k}: {distance}")
```
