```python
import numpy as np
```

```python
docs= ["The cow jumped over the moon"
,"O'Leary's cow kicked the lamp"
,"The kicked lamp started a fire"
,"The cow on fire"]
print(docs[1])
```

```python
vocb = {}

for d in docs:
    lowered= d.lower()
    tok = lowered.split()
    for key in tok:
        ret = vocb.get(key, 0)
        if ret == 0:
            vocb.update({key : 1})
        else:
            vocb[key]+=1
print(vocb) 
```
