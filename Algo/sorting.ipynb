```python
def merge(arr1, arr2):
    # print(f"in merge!! arr1 {arr1}, arr2 {arr2}")
    arr= []
    i=0
    j=0
    # print(f"going in while, i is {i} j:{j}")
    # print("lengths are", len(arr2), len(arr1))
    while j<len(arr2) and i<len(arr1):
        # print(f"i is {i}, j is {j}")
        if arr1[i]>arr2[j]:
            arr.append( arr2[j])
            j+=1
        else:
            arr.append(arr1[i])
            i+=1
    # print("out of while")
    if j<len(arr2):
        arr.extend(arr2[j:])
    if i<len(arr1):
            arr.extend(arr1[i:])
    # print(f"in merge: {arr}")
    return arr
```
