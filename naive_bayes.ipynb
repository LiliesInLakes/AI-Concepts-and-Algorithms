```python
import numpy as numpy
import matplotlib as plt
import pandas as pd
```

```python
df= pd.read_csv("/home/student/Downloads/assignment09_flu_symptoms.csv")
print(df.head())
print(df.columns.tolist())
print(df.shape)
```

```python
total_patients= len(df)

flue_counts= df["Flu"].value_counts()
pFy= flue_counts[0]/total_patients
pFn= flue_counts[1]/total_patients
print("Prior classes:", pFy, pFn)

results= []

for symp in df.columns.tolist():
    if symp == "Flu" or symp == "Headache":
        continue
    pyy= ((df[symp]== "Y") & (df["Flu"]== "Y")).sum()/ flue_counts[0]
    pny= ((df[symp]== "N") & (df["Flu"]== "Y")).sum()/ flue_counts[0]
    pyn= ((df[symp]== "Y") & (df["Flu"]== "N")).sum()/ flue_counts[1]
    pnn= ((df[symp]== "N") & (df["Flu"]== "N")).sum()/ flue_counts[1]
    results.append([
        symp, pyy, pny, pyn, pnn,
        None, None, None, None, None, None
    ])

# table= pd.DataFrame(
#     results, 
#     columns= ["Symptom",
#               "P(Sym=Y| FLu=Y)",
#               "P(Sym=N| FLu=Y)",
#               "P(Sym=Y| FLu=N)",
#               "P(Sym=N| FLu=N)"]
# )

# print(table.round(3).to_string(index=False))

headche=["Headache", None, None, None, None]
headche.append(((df["Headache"]=="Strong") & (df["Flu"]=="Y")).sum()/flue_counts[0])
headche.append(((df["Headache"]=="Mild") & (df["Flu"]=="Y")).sum()/flue_counts[0])
headche.append(((df["Headache"]=="No") & (df["Flu"]=="Y")).sum()/flue_counts[0])
headche.append(((df["Headache"]=="Strong") & (df["Flu"]=="N")).sum()/flue_counts[1])
headche.append(((df["Headache"]=="Mild") & (df["Flu"]=="N")).sum()/flue_counts[1])
headche.append(((df["Headache"]=="No") & (df["Flu"]=="N")).sum()/flue_counts[1])

results.append(headche)
table2= pd.DataFrame(
    results, 
    columns= ["Symptom",
              "P(Y|Flu=Y)",
              "P(N|Flu=Y)",
              "P(Y|Flu=N)",
              "P(N|Flu=N)",
              "P(Strong|Flu=Y)", "P(Mild|Flu=Y)", "P(No|Flu=Y)",
    "P(Strong|Flu=N)", "P(Mild|Flu=N)", "P(No|Flu=N)"]
)

print(table2.round(3).to_string(index=False))
```

```python
sick= {
    "Chills": "Y",
    "RunnyNose": "N",
    "Headache": "Mild",
    "Fever": "Y"
}

def calc_score(sick, flue_clss):
    print(f"For Flu= {flue_clss}")
    if flue_clss == "Y":
        prob= pFy
    else:
        prob= pFn
    for symp, vlue in sick.items():
        print(f"    For symptom: {symp}")
        print("     ", round(prob, 2))
        prob_temp= table2.loc[table2["Symptom"]== symp , f"P({vlue}|Flu={flue_clss})"].iloc[0]
        prob= prob*prob_temp
    return prob

sick_y= calc_score(sick, "Y")

sick_n= calc_score(sick, "N")
print(sick_y, sick_n)
if(sick_n>sick_y):
    print("No flu")
else:
    print("Yes flu")
      
```
